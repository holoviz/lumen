# Ask your data a question from a notebook

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/holoviz/lumen/blob/main/examples/notebooks/ask_in_a_notebook.ipynb)

Load data with `lumen.read`, then ask it questions in plain language. Each answer comes back with the SQL that produced it, a chart when one helps, and code that reruns it without an LLM. Answers depend on the model, so your wording and charts may differ from run to run.

```python
# In Colab, or anywhere Lumen is not installed yet
%pip install -q "lumen[ai-openai]" hvsampledata
```

```python
import hvsampledata
import lumen

from lumen.ai.llm import Kilo

penguins = hvsampledata.penguins("pandas")  # ships with the package, no download
source = lumen.read({"penguins": penguins})

# Kilo serves free models without an account. With an API key set,
# e.g. OPENAI_API_KEY, leave out llm= and Lumen picks that provider.
llm = Kilo()
```

## Ask a question

```python
answer = source.ask("Which species is heaviest on average?", llm=llm)
answer
```

Every answer keeps the query and the data behind it.

```python
print(answer.sql)
answer.pipeline.data
```

## Ask for a chart

```python
source.ask("Plot the average body mass per species as a bar chart", llm=llm)
```

## Follow up

A chat remembers the conversation, and each question runs on the data the last answer left.

```python
chat = source.chat(llm=llm)
chat.ask("What is the average body mass per species?")
```

```python
chat.ask("Only keep the species above 4000 g")
```

## Rerun without an LLM

`answer.code` rebuilds the result with Lumen alone. For large data, `answer.export(folder)` writes a small script and the data as Parquet files.

```python
print(answer.code[:400])
```

## What Lumen runs

Lumen only writes SQL and declarative chart specs, and shows both to you. It never runs Python written by the LLM unless you turn that on, so asking a question cannot execute arbitrary code on your machine.
